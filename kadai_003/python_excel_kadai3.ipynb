```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill

# 年間売上表の読み込み
df_2022 = pd.read_excel('2022_年間売上表.xlsx')
df_2023 = pd.read_excel('2023_年間売上表.xlsx')

# データの連結と集計
# データの連結 (縦にくっつける)
df_concat = pd.concat([df_2022, df_2023])

# 「商品」「売上年」ごとにグループ化し、合計金額に集約
# as_index=False にすることで、マルチインデックス化を阻止
df_grouped = df_concat.groupby(['商品', '売上年'], as_index=False)['金額（千円）'].sum()

# 売上集計表の作成とヘッダーの書式変更
output_file = '売上集計表.xlsx'

# pandasで集計結果を一度Excelファイルとして保存する
df_grouped.to_excel(output_file, index=False)

# 保存したファイルをopenpyxlで読み込み、見た目を調整する
wb = load_workbook(output_file)
ws = wb.active

# 塗りつぶしの色を設定（薄いグレー: F2F2F2）
gray_fill = PatternFill(patternType='solid', fgColor='F2F2F2')

# 1行目（ヘッダー）のセルを順番に取り出し、背景色を適用する
for cell in ws[1]:
    cell.fill = gray_fill

# 最後に上書き保存
wb.save(output_file)
```
