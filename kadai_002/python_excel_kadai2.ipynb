```python
import pandas as pd

# 1. 元となるデータの作成
data = {
    '日付': ['2023-05-17', '2023-05-18', '2023-05-19', '2023-05-20', '2023-05-21'],
    '社員名': ['山田', '佐藤', '鈴木', '田中', '高橋'],
    '売上': [100, 200, 150, 300, 250],
    '部門': ['メーカー', '代理店', 'メーカー', '商社', '代理店']
}
df = pd.DataFrame(data)

# 2. 「平均売上」列の作成
# mean() を使って売上の平均値を算出し、新しい列に代入します
df['平均売上'] = df['売上'].mean()

# 3. 関数「performance」の定義
# 行(row)ごとにデータを受け取り、条件分岐でランクを返します
def performance(row):
    sales = row['売上']
    avg_sales = row['平均売上']

    if sales >= avg_sales + 50:    # 平均売上+50以上の場合
        return 'A'
    elif sales >= avg_sales:       # 平均売上以上の場合
        return 'B'
    else:                          # 平均売上未満の場合
        return 'C'

# 4. 「業績ランク」列の作成
# apply関数に axis=1 を指定することで、1行ずつ関数にデータを渡して処理できます
df['業績ランク'] = df.apply(performance, axis=1)

# 5. 新規Excelファイルへの書き出し
# index=False を指定することで、一番左の連番（インデックス）が出力されるのを防ぎます
df.to_excel('業績.xlsx', index=False)
```
