```python
import openpyxl
from datetime import datetime

# 1. 現在日付を取得し、保存するファイル名を作成する
today = datetime.now()
date_str = today.strftime('%Y%m%d') # 例: 20230520
file_name = f"請求書_{date_str}.xlsx"

# 2. Excelワークブックの新規作成
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "請求書" # シート名を変更（任意）

# 3. データをセルに入力していく
ws['B2'] = "請求書"

# 宛先・伝票情報
ws['B4'] = "株式会社ABC"
ws['F4'] = "No."
ws['G4'] = "0001"  # 文字列として入力（数値の1にしない）
ws['B5'] = "〒101-0022 東京都千代田区神田練塀町300"
ws['F5'] = "日付"
ws['G5'] = today.strftime('%Y/%m/%d')
ws['B6'] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws['B7'] = "担当者名:鈴木一郎 様"

# 明細行のヘッダー
ws['B10'] = "商品名"
ws['C10'] = "数量"
ws['D10'] = "単価"
ws['E10'] = "金額"

# 明細データ（商品A）
ws['B11'] = "商品A"
ws['C11'] = 2
ws['D11'] = 10000
ws['E11'] = 20000

# 明細データ（商品B）
ws['B12'] = "商品B"
ws['C12'] = 1
ws['D12'] = 15000
ws['E12'] = 15000

# 合計金額等
ws['B15'] = "小計"
ws['E15'] = 35000
ws['B16'] = "消費税"
ws['E16'] = 3500
ws['B17'] = "合計"
ws['E17'] = 38500

# 4. ファイルを保存する
wb.save(file_name)
print(f"{file_name} を作成しました！")
```
